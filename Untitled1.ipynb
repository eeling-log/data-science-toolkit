```python
import pandas as pd
import random

# Generate 1,000 rows of synthetic data
data = []
for _ in range(15000):
    has_cancer = random.random() > 0.7  # 30% cancer cases
    cancer_type = random.choice(["lung cancer", "tracheal cancer", "bronchial cancer", "none"]) if has_cancer else "none"

    data.append({
        "height_cm": random.randint(150, 190),
        "weight_kg": random.randint(45, 100),
        "occupation": random.choice(["Office worker", "Construction", "Healthcare", "Driver", "Factory"]),
        "exercise": random.choice(["never", "sometimes", "often"]),
        "sleep_hours": random.randint(4, 10),
        "stay_up_late": random.choice(["never", "sometimes", "often"]),
        "ever_smoked": "yes" if random.random() > 0.5 else "no",
        "cigarettes_per_day": random.randint(0, 30) if has_cancer else random.randint(0, 10),
        "smoking_years": random.randint(0, 40) if has_cancer else random.randint(0, 15),
        "quit_smoking": random.choice(["yes", "no"]),
        "secondhand_smoke": random.choice(["never", "sometimes", "often"]),
        "tobacco_use": random.choice(["never", "sometimes", "often"]),
        "alcohol_use": random.choice(["never", "sometimes", "often"]),
        "fruits_veggies": random.choice(["never", "sometimes", "often"]),
        "processed_meat": random.choice(["never", "sometimes", "often"]),
        "respiratory_disease": "yes" if has_cancer else random.choice(["yes", "no"]),
        "dust_fumes_exposure": random.choice(["never", "sometimes", "often"]),
        "asbestos_chemical_exposure": "often" if has_cancer else random.choice(["never", "sometimes"]),
        "family_cancer_history": random.choice(["none", "lung cancer"]),
        "diagnosis": cancer_type
    })

pd.DataFrame(data).to_csv("respiratory_cancer_data.csv", index=False)
print("Dataset created as 'respiratory_cancer_data.csv'!")
```
