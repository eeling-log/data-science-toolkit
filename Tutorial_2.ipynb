```python
#Part 1-Google Colab and Python
#1. Create a new notebook and code the following
#1. Create a numpy array of size 2 x 2

import numpy as np

array_2x2 =np.array([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4
#The first number is the row and second number is the column

random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes

ones_array = np.ones((3,3))
zeroes_array = np.zeros((3,3))

print("Array of Ones: ")
print(ones_array)

print("Array of zeroes: ")
print(zeroes_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it

random6x3 = np.random.rand(6,3)

print("Original array: ")
print(random6x3)

transposed_array = random6x3.T
print("Transposed array: ")
print(transposed_array)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product

arr1 = np.random.rand(1,2)
arr2 = np.random.rand(2,3)

print("Array 1: ")
print(arr1)

print("Array 2: ")
print(arr2)

dot_product = np.dot(arr1, arr2)

print("Result of dot product:")
print(dot_product)
```
